```python
# Draw a Ball and Background
# sky gradient-ish background
for y in range(height):
    r = 26 + int((y / height) * 35)
    g = 38 + int((y / height) * 48)
    b = 58 + int((y / height) * 60)
    canvas.line((0, y, width, y), fill=(r, g, b, 255))

# ground plane
canvas.rounded_rectangle((0, 360, width, height), radius=26, fill=(45, 74, 60, 255))

# ball shadow
shadow = (180, 210, 230, 120)
canvas.ellipse((150, 300, 360, 350), fill=shadow)

# main ball orb
ball_x, ball_y = 256, 210
ball_r = 72
ball_color = (220, 238, 255, 255)
canvas.ellipse((ball_x - ball_r, ball_y - ball_r, ball_x + ball_r, ball_y + ball_r), fill=ball_color)

# ball highlight
canvas.ellipse((ball_x - 34, ball_y - 42, ball_x + 18, ball_y + 4), fill=(255, 255, 255, 170))

# small rim detail on ball
canvas.ellipse((ball_x - ball_r, ball_y - ball_r, ball_x + ball_r, ball_y + ball_r), outline=(157, 177, 220, 220), width=6)

# subtle glow behind ball
for glow in [92, 110, 130]:
    canvas.ellipse((ball_x - glow, ball_y - glow, ball_x + glow, ball_y + glow), outline=(135, 180, 255, 70), width=2)

img = img.filter(ImageFilter.GaussianBlur(0.8))
```
